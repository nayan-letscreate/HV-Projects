# Task 1: For the suburb of Altona, it is postulated that a typical property sells for $800,000. 

##### 1. Use the data at hand to test this assumption for significance level of 5%
##### 2. Is the typical property price really $800,000 or has it increased?
Response: There is insufficient evidence that the price has increased beyond $800,000 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv("D:\Documents\Hero Vired\Python for Data Analysis\Assignment 3\property.csv")
```

```python
df.head()
```

```python
altona = df[df['Suburb'] == 'Altona']['Price'].dropna()

print("=" * 50)
print("TASK 1 — One Sample Hypothesis Test (Altona)")
print("=" * 50)
```

```python
# ── Descriptive Stats ──────────────────────────────────────
print(f"\nSample Size        : {len(altona)}")
print(f"Sample Mean        : ${altona.mean():,.2f}")
print(f"Sample Median      : ${altona.median():,.2f}")
print(f"Std Deviation      : ${altona.std():,.2f}")
```

```python
# ── Hypotheses ─────────────────────────────────────────────
# H0 : μ = 800,000  (typical price is $800,000)
# H1 : μ > 800,000  (price has increased — one-tailed)

pop_mean = 800000
alpha    = 0.05

print(f"\nH0 : Mean price = $800,000")
print(f"H1 : Mean price > $800,000  (one-tailed, right)")
print(f"Significance Level : {alpha}")
```

```python
# ── One Sample T-Test ──────────────────────────────────────
t_stat, p_two_tailed = stats.ttest_1samp(altona, popmean=pop_mean)
```

```python
# One-tailed p-value (right tail since H1 is μ > 800,000)
p_one_tailed = p_two_tailed / 2 if t_stat > 0 else 1 - (p_two_tailed / 2)

print(f"\nT-Statistic        : {t_stat:.4f}")
print(f"P-Value (one-tail) : {p_one_tailed:.4f}")
```

```python
# ── Decision ───────────────────────────────────────────────
print("\n" + "-" * 50)
if p_one_tailed < alpha:
    print("DECISION : Reject H0")
    print("CONCLUSION : There is sufficient evidence that the")
    print("             typical property price in Altona has")
    print("             INCREASED beyond $800,000.")
else:
    print("DECISION : Fail to Reject H0")
    print("CONCLUSION : There is insufficient evidence that")
    print("             the price has increased beyond $800,000.")
print("-" * 50)
```

# Task 2:  For the year 2016, is there any difference in the prices of properties sold in the summer months vs winter months?

Response: There IS a statistically significant difference in property prices between summer and winter months in 2016. Winter prices are HIGHER than summer.


```python
# ── Parse Dates ─────────────────────────────────────
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
df['Year']  = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
```

```python
# ── Filter for 2016 ────────────────────────────────────────
df_2016 = df[df['Year'] == 2016][['Month', 'Price']].dropna()
```

```python
# ── Assign Season ──────────────────────────────────────────
# Winter : Oct(10), Nov(11), Dec(12), Jan(1), Feb(2), Mar(3)
# Summer : Apr(4), May(5), Jun(6), Jul(7), Aug(8), Sep(9)

winter_months = [10, 11, 12, 1, 2, 3]
summer_months = [4, 5, 6, 7, 8, 9]

winter = df_2016[df_2016['Month'].isin(winter_months)]['Price']
summer = df_2016[df_2016['Month'].isin(summer_months)]['Price']

print("=" * 55)
print("TASK 2 — Two Sample T-Test (Summer vs Winter 2016)")
print("=" * 55)
```

```python
# ── Descriptive Stats ──────────────────────────────────────
print(f"\n{'Metric':<25} {'Winter':>15} {'Summer':>15}")
print("-" * 55)
print(f"{'Sample Size':<25} {len(winter):>15} {len(summer):>15}")
print(f"{'Mean Price':<25} ${winter.mean():>14,.2f} ${summer.mean():>14,.2f}")
print(f"{'Median Price':<25} ${winter.median():>14,.2f} ${summer.median():>14,.2f}")
print(f"{'Std Deviation':<25} ${winter.std():>14,.2f} ${summer.std():>14,.2f}")
```

```python
# ── Hypotheses ─────────────────────────────────────────────
# H0 : μ_winter = μ_summer  (no difference in prices)
# H1 : μ_winter ≠ μ_summer  (two-tailed — any difference)

alpha = 0.05
print(f"\nH0 : Mean price (Winter) = Mean price (Summer)")
print(f"H1 : Mean price (Winter) ≠ Mean price (Summer)  (two-tailed)")
print(f"Significance Level : {alpha}")
```

```python
# ── Levene's Test for Equal Variance ──────────────────────
lev_stat, lev_p = stats.levene(winter, summer)
print(f"\nLevene's Test p-value : {lev_p:.4f}")
equal_var = lev_p > 0.05
print(f"Equal Variances Assumed : {equal_var}")
```

```python
# ── Independent Two Sample T-Test ─────────────────────────
t_stat, p_value = stats.ttest_ind(winter, summer, equal_var=equal_var)

print(f"\nT-Statistic : {t_stat:.4f}")
print(f"P-Value     : {p_value:.4f}")
```

```python
# ── Decision ───────────────────────────────────────────────
print("\n" + "-" * 55)
if p_value < alpha:
    print("DECISION   : Reject H0")
    print("CONCLUSION : There IS a statistically significant")
    print("             difference in property prices between")
    print("             summer and winter months in 2016.")
    if winter.mean() > summer.mean():
        print("             Winter prices are HIGHER than summer.")
    else:
        print("             Summer prices are HIGHER than winter.")
else:
    print("DECISION   : Fail to Reject H0")
    print("CONCLUSION : There is NO statistically significant")
    print("             difference in property prices between")
    print("             summer and winter months in 2016.")
print("-" * 55)
```
