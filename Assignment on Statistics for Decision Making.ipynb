# Task 1: For the suburb of Altona, it is postulated that a typical property sells for $800,000. 

##### 1. Use the data at hand to test this assumption for significance level of 5%
##### 2. Is the typical property price really $800,000 or has it increased?
Response: There is insufficient evidence that the price has increased beyond $800,000 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv("D:\Documents\Hero Vired\Python for Data Analysis\Assignment 3\property.csv")
```

```python
df.head()
```

```python
altona = df[df['Suburb'] == 'Altona']['Price'].dropna()

print("=" * 50)
print("TASK 1 — One Sample Hypothesis Test (Altona)")
print("=" * 50)
```

```python
# ── Descriptive Stats ──────────────────────────────────────
print(f"\nSample Size        : {len(altona)}")
print(f"Sample Mean        : ${altona.mean():,.2f}")
print(f"Sample Median      : ${altona.median():,.2f}")
print(f"Std Deviation      : ${altona.std():,.2f}")
```

```python
# ── Hypotheses ─────────────────────────────────────────────
# H0 : μ = 800,000  (typical price is $800,000)
# H1 : μ > 800,000  (price has increased — one-tailed)

pop_mean = 800000
alpha    = 0.05

print(f"\nH0 : Mean price = $800,000")
print(f"H1 : Mean price > $800,000  (one-tailed, right)")
print(f"Significance Level : {alpha}")
```

```python
# ── One Sample T-Test ──────────────────────────────────────
t_stat, p_two_tailed = stats.ttest_1samp(altona, popmean=pop_mean)
```

```python
# One-tailed p-value (right tail since H1 is μ > 800,000)
p_one_tailed = p_two_tailed / 2 if t_stat > 0 else 1 - (p_two_tailed / 2)

print(f"\nT-Statistic        : {t_stat:.4f}")
print(f"P-Value (one-tail) : {p_one_tailed:.4f}")
```

```python
# ── Decision ───────────────────────────────────────────────
print("\n" + "-" * 50)
if p_one_tailed < alpha:
    print("DECISION : Reject H0")
    print("CONCLUSION : There is sufficient evidence that the")
    print("             typical property price in Altona has")
    print("             INCREASED beyond $800,000.")
else:
    print("DECISION : Fail to Reject H0")
    print("CONCLUSION : There is insufficient evidence that")
    print("             the price has increased beyond $800,000.")
print("-" * 50)
```

# Task 2:  For the year 2016, is there any difference in the prices of properties sold in the summer months vs winter months?

Response: There IS a statistically significant difference in property prices between summer and winter months in 2016. Winter prices are HIGHER than summer.


```python
# ── Parse Dates ─────────────────────────────────────
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
df['Year']  = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
```

```python
# ── Filter for 2016 ────────────────────────────────────────
df_2016 = df[df['Year'] == 2016][['Month', 'Price']].dropna()
```

```python
# ── Assign Season ──────────────────────────────────────────
# Winter : Oct(10), Nov(11), Dec(12), Jan(1), Feb(2), Mar(3)
# Summer : Apr(4), May(5), Jun(6), Jul(7), Aug(8), Sep(9)

winter_months = [10, 11, 12, 1, 2, 3]
summer_months = [4, 5, 6, 7, 8, 9]

winter = df_2016[df_2016['Month'].isin(winter_months)]['Price']
summer = df_2016[df_2016['Month'].isin(summer_months)]['Price']

print("=" * 55)
print("TASK 2 — Two Sample T-Test (Summer vs Winter 2016)")
print("=" * 55)
```

```python
# ── Descriptive Stats ──────────────────────────────────────
print(f"\n{'Metric':<25} {'Winter':>15} {'Summer':>15}")
print("-" * 55)
print(f"{'Sample Size':<25} {len(winter):>15} {len(summer):>15}")
print(f"{'Mean Price':<25} ${winter.mean():>14,.2f} ${summer.mean():>14,.2f}")
print(f"{'Median Price':<25} ${winter.median():>14,.2f} ${summer.median():>14,.2f}")
print(f"{'Std Deviation':<25} ${winter.std():>14,.2f} ${summer.std():>14,.2f}")
```

```python
# ── Hypotheses ─────────────────────────────────────────────
# H0 : μ_winter = μ_summer  (no difference in prices)
# H1 : μ_winter ≠ μ_summer  (two-tailed — any difference)

alpha = 0.05
print(f"\nH0 : Mean price (Winter) = Mean price (Summer)")
print(f"H1 : Mean price (Winter) ≠ Mean price (Summer)  (two-tailed)")
print(f"Significance Level : {alpha}")
```

```python
# ── Levene's Test for Equal Variance ──────────────────────
lev_stat, lev_p = stats.levene(winter, summer)
print(f"\nLevene's Test p-value : {lev_p:.4f}")
equal_var = lev_p > 0.05
print(f"Equal Variances Assumed : {equal_var}")
```

```python
# ── Independent Two Sample T-Test ─────────────────────────
t_stat, p_value = stats.ttest_ind(winter, summer, equal_var=equal_var)

print(f"\nT-Statistic : {t_stat:.4f}")
print(f"P-Value     : {p_value:.4f}")
```

```python
# ── Decision ───────────────────────────────────────────────
print("\n" + "-" * 55)
if p_value < alpha:
    print("DECISION   : Reject H0")
    print("CONCLUSION : There IS a statistically significant")
    print("             difference in property prices between")
    print("             summer and winter months in 2016.")
    if winter.mean() > summer.mean():
        print("             Winter prices are HIGHER than summer.")
    else:
        print("             Summer prices are HIGHER than winter.")
else:
    print("DECISION   : Fail to Reject H0")
    print("CONCLUSION : There is NO statistically significant")
    print("             difference in property prices between")
    print("             summer and winter months in 2016.")
print("-" * 55)
```

# Task 3: For the suburb of Abbotsford, what is the probability that out of 10 properties sold, 3 will not have a car parking space?
Response: The probability that exactly 3 out of 10 properties will NOT have a car parking space is 0.26

```python
# ── Filter for Abbotsford ──────────────────────────────────
abbotsford = df[df['Suburb'] == 'Abbotsford']['Car'].dropna()

print("=" * 55)
print("TASK 3 — Binomial Probability (Abbotsford)")
print("=" * 55)
```

```python
# ── Step 1: Find P(no car parking) ────────────────────────
no_car  = (abbotsford == 0).sum()   # properties with car == 0
total   = len(abbotsford)
p_no_car = no_car / total

print(f"\nTotal Properties (Abbotsford) : {total}")
print(f"Properties with No Parking    : {no_car}")
print(f"P(no car parking)             : {p_no_car:.4f}")
```

```python
# ── Step 2: Binomial Setup ─────────────────────────────────
# X ~ Binomial(n=10, p=p_no_car)
# Find P(X = 3)

n = 10   # total properties
k = 3    # properties with no parking
 
print(f"\nBinomial Parameters:")
print(f"  n (trials)   : {n}")
print(f"  k (success)  : {k}  (no car parking)")
print(f"  p (prob)     : {p_no_car:.4f}")
```

```python
# ── Step 3: Calculate Probability ─────────────────────────
# P(X=3) = C(10,3) * p^3 * (1-p)^7

prob = stats.binom.pmf(k=k, n=n, p=p_no_car)

print(f"\nP(X = 3) using stats.binom.pmf : {round(prob, 3)}")
```

```python
from math import comb
manual = comb(n, k) * (p_no_car**k) * ((1 - p_no_car)**(n - k))
print(f"P(X = 3) manual calculation   : {round(manual, 3)}")
```

```python
# ── Final Answer ───────────────────────────────────────────
print("\n" + "-" * 55)
print(f"ANSWER : The probability that exactly 3 out of 10")
print(f"         properties will NOT have a car parking")
print(f"         space is : {round(prob, 3)}")
print("-" * 55)
```

# Task 4: In the suburb of Abbotsford, what are the chances of finding a property with 3 rooms? Round your answer to 3 decimal places. 
Response: The probability of finding a property with 3 rooms in Abbotsford is 0.357

```python
# ── Step 1: P(Rooms == 3) ──────────────────────────────────
total        = len(df[df['Suburb'] == 'Abbotsford']['Rooms'])
count_3rooms = (df[df['Suburb'] == 'Abbotsford']['Rooms'] == 3).sum()
p_3rooms     = count_3rooms / total

print(f"\nTotal Properties    : {total}")
print(f"Properties (3 rooms): {count_3rooms}")
```

```python
# ── Frequency Distribution ─────────────────────────────────
room_counts = df[df['Suburb'] == 'Abbotsford']['Rooms'].dropna().value_counts().sort_index()
room_pct    = df[df['Suburb'] == 'Abbotsford']['Rooms'].dropna().value_counts(normalize=True).sort_index()

print("\nRoom Distribution in Abbotsford:")
print(f"\n{'Rooms':<10} {'Count':>10} {'Probability':>15}")
print("-" * 38)
for room in room_counts.index:
    print(f"{int(room):<10} {int(room_counts[room]):>10} {round(room_pct[room], 3):>15}")
```

```python
# ── Final Answer ───────────────────────────────────────────
print("\n" + "-" * 55)
print(f"ANSWER : The probability of finding a property")
print(f"         with 3 rooms in Abbotsford is : {round(p_3rooms, 3)}")
print("-" * 55)
```

# Task 6: One-Sample Hypothesis Test (Industry Pricing) A real estate firm claims that the average property price in Richmond is $1,000,000. Using the dataset, test whether the actual average price is significantly different from this claim at a 5% significance level. Clearly state: 
• Null and alternative hypotheses 
• Test statistic 
• p-value 
• Final business conclusion 

```python
richmond = df[df['Suburb'] == 'Richmond']['Price'].dropna()

print("=" * 55)
print("TASK 6 — One Sample T-Test (Richmond)")
print("=" * 55)
```

```python
# ── Descriptive Stats ──────────────────────────────────────
print(f"\nSample Size   : {len(richmond)}")
print(f"Sample Mean   : ${richmond.mean():,.2f}")
print(f"Std Deviation : ${richmond.std():,.2f}")
```

```python
# ── Hypotheses ─────────────────────────────────────────────
# H0 : μ = 1,000,000  (average price is $1M)
# H1 : μ ≠ 1,000,000  (average price is different — two-tailed)

pop_mean = 1000000
alpha    = 0.05

print(f"\nH0 : Mean price = $1,000,000")
print(f"H1 : Mean price ≠ $1,000,000  (two-tailed)")
print(f"Significance Level : {alpha}")
```

```python
# ── One Sample T-Test ──────────────────────────────────────
t_stat, p_value = stats.ttest_1samp(richmond, popmean=pop_mean)

print(f"\nT-Statistic : {t_stat:.4f}")
print(f"P-Value     : {p_value:.4f}")
```

```python
# ── Decision ───────────────────────────────────────────────
print("\n" + "-" * 55)
if p_value < alpha:
    print("DECISION   : Reject H0")
    print(f"CONCLUSION : The actual average price in Richmond")
    print(f"             is significantly DIFFERENT from $1,000,000.")
    direction = "ABOVE" if richmond.mean() > pop_mean else "BELOW"
    print(f"             The true mean appears to be {direction} the claim.")
else:
    print("DECISION   : Fail to Reject H0")
    print(f"CONCLUSION : There is insufficient evidence that the")
    print(f"             average price differs from $1,000,000.")
print("-" * 55)
```
